 ***Soru 1*** Klavyeden girilen boy bilgisine göre kısa, normal veya uzun çıktısı veren programı yaz

```python
# Klavyeden girilen boy bilgisine göre kısa, normal veya uzun çıktısı veren programı yaz
boy = int(input("Boyunuzu cm olarak giriniz : "))
if boy < 160 :
  print("Kısa Boylusunuz")
elif 160 <= boy < 180 : # boy < 180 # boy => 160 and boy < 180
  print("Orta Boylusunuz")
else:
  print("Uzun Boylusunuz")
```

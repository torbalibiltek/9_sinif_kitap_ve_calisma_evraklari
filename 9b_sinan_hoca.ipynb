```python
"""
Bir hava yolu firması en fazla 20 kilogram bagaj hakkı vermektedir. 20 kilogramdan sonraki her kilogram
için 10 TL ek ücret almaktadır. Buna göre bagajı 20 kg ya da daha az olan yolculara “Herhangi bir ücret
ödemeniz gerekmiyor.”; 20 kg’den fazla olanlar için de ne kadar ek ücret ödeneceğini hesaplayarak “Fazla
bagaj için ….. TL ödemelisiniz.” çıktılarını veren kodu yazınız.
Not: Bu soruda kilogram hesabında sadece tam sayıları dikkate alınız. Örneğin 28,70 kilogram olan bagaj
için sadece 8 kg için ek ücret ödenmesi yeterlidir.
"""

bavul_agirligi = int(float(input("bavul ağırlığını gir")))
if bavul_agirligi <= 20:
  print("ek ödeme yok")
else:
  fark = bavul_agirligi - 20
  tutar = fark * 10
  print(f"bavulunuz {fark}kg fazla {tutar}TL ödemelisiniz")

```

```python
"""
Girilen iki sayıya ve operatöre (+,-,*,/) göre toplama, çıkarma, çarpma ya da bölme işlemlerini
yapan; bu operatörler dışında bir değer girildiğinde “Yanlış işlem girdiniz.” uyarısı veren kodu yazınız
"""
sayi1 = int(input("sayı giriniz"))
sayi2 = int(input("sayı giriniz"))
islem = input("işlem giriniz + - / * % // **")
durum = True
if islem == "+":
  sonuc = sayi1 + sayi2
elif islem == "-":
  sonuc =  sayi1 - sayi2
elif islem == "/":
  sonuc =  sayi1 / sayi2
elif islem == "*":
  sonuc =  sayi1 * sayi2
elif islem == "%":
  sonuc =  sayi1 % sayi2
elif islem == "//":
  sonuc =  sayi1 // sayi2
elif islem == "**":
  sonuc =  sayi1 ** sayi2
else:
  print("Hatalı İşlem")
  durum = False


if durum:
  print(f"{sayi1} {islem} {sayi2} = {sonuc}")
```

```python
"""
Kullanıcının girdiği boy ve ağırlık değerlerine göre vücut kitle indeksini (VKİ=ağırlık/(boy*boy), boy
metre cinsinden verilmeli) hesaplayınız.
VKİ 18 ile < 25 aralığındaysa normal,
VKİ 25 ile <30 aralığındaysa kilolu,
VKİ 30 ve daha yüksekse obez,
VKİ 35 ve daha fazlaysa ciddi obez olarak kabul edilir.
VKİ’ni hesaplayarak kişinin durumunu yazdırınız.
"""
kilo = float(input("Kilo giriniz"))
boy = float(input("Boy giriniz"))
vki = kilo/(boy**2)
if vki <18 :
  print("zayıf")
elif 18 <= vki < 25:
  print("normal")
elif 18 <= vki < 25:
  print("normal")
elif 25 <= vki < 30:
  print("kilolu")
elif 30 <= vki < 35:
  print("obez")
else:
  print("aşırı obez")
```

```python
from datetime import datetime
sorular = {"soru1":{"soru":"birinci soru","siklar":["a","b","c"],"cevap":"a"},
           "soru2":{"soru":"ikinci soru","siklar":["a","b","c"],"cevap":"c"},
           "soru3":{"soru":"ücüncü soru","siklar":["a","b","c"],"cevap":"b"}}

def sinav(sorular):
  sonuclar = {"dogru":0, "yanlis":0, "bos":0}
  for soru in sorular:
    print(sorular[soru]['soru'])
    siklar = sorular[soru]['siklar']
    print(siklar)
    baslangic = datetime.now()
    cevap = input("yukarıdaki soruyu cevaplayın")
    bitis = datetime.now()
    fark = bitis - baslangic
    print("toplam cevap verme süresi", fark.seconds)
    if fark.seconds <= 60:
      if cevap == sorular[soru]['cevap']:
        print("Bildiniz")
        sonuclar["dogru"] += 1
      else:
        print("Bilemediniz")
        sonuclar["yanlis"] += 1
    else:
      print("Süre Bitti")
      sonuclar["bos"] += 1

  return sonuclar


degerlendirme = sinav(sorular)
print("Sınav Sonucunuz Değerlendirmesi", degerlendirme)
```

```python
#şanslı yedili
"""
0 9 0
7 7 7 ++
1 2 3
4 4 4 +
"""
from random import randint
def SansliYediliOyna(jeton, amorti=0):
  if jeton == 1 :
    oyun = 1
  elif jeton == 5:
    oyun = 10
  elif jeton == 10:
    oyun = 250
  else:
    oyun = 1

  if amorti > 0:
    oyun = amorti
    amorti = 0
  for dd in range(oyun):
    a = randint(0,9)
    b = randint(0,9)
    c = randint(0,9)
    print(a,b,c)
    if a == b == c == 7:
      print("Kazandınız 100 puan")
      #print(a,b,c)
      #break
      return True
    elif a == b == c:
      print("Kazandınız 50 puan")
      #print(a,b,c)
      #break
      return False
    else:
      print("Kaybettiniz")
    if a== b or b == c or a == c:
      print("Amortiiiiii")
      amorti += 1

  if amorti > 0:
    print("Amortiler Oynanıyor")
    return SansliYediliOyna(1, amorti)
SansliYediliOyna(10)
```

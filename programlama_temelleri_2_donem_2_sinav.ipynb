# **Soru 1**
5 adet meyveden oluşan bir liste tanımlayınız ve bu listeyi bir döngü aracılığı ile yazdırınız

```python
meyveler = ["karpuz","çilek","şeftali","erik","kiraz"]
#8 puan yukarısı

#8 puan aşağısı
for meyve in meyveler:
  print(meyve)

index = 0
while index < len(meyveler):
  print(meyveler[index])
  index += 1
```

# **Soru 2**
Yukarıdaki listeye yeni bir meyve ekleyiniz. Meyvenin listede varlığını kontrol ederek, meyve listede varsa "Listede bu meyve var", yoksa listeye meyveyi ekleyerek listeyi yazdırın.

```python
meyve = input("listeye eklenecek meyve giriniz")
#meyve = "portakal"
#8 puan if
if meyve in meyveler:
  print("Meyve Listede Var")
else: #8 puan else
  meyveler.append(meyve)
  print(meyveler)
```

# **Soru 3**
Öğrenciler ve ortalamaları ola 5 kişilik bir ogrenci_puanlari adında bir sözlük tanımlayınız. öğrenci ve puanlarını bir döngü ile yazdırın ayrıca öğrencilerin genel ortalamasını yazdırın

```python
ogrenci_puanlari = {"a":50, "b":60 , "c":70,
                    "d":60, "e": 80}
#sözlük tanımlamak 8 puan
#sözlüğü yazdırmak 5 puan
#ortalamayı hesaplatıp yazdırmak 3 puan
toplam = 0
for ogrenci in ogrenci_puanlari:
  print(ogrenci,"=>",ogrenci_puanlari[ogrenci])
  toplam += ogrenci_puanlari[ogrenci]
ortalama = toplam / len(ogrenci_puanlari)
print("Genel Ortalama : ", ortalama)
```

# **Soru 4**
vki hesapla ve değerlendir

```python
agirlik = float(input("agirlik bilgisi giriniz"))
boy = float(input("boy bilgisi giriniz"))
#bilgileri almak 4 puan
vki = agirlik / (boy * boy)
#hesaplamayı yapmak 4 puan
print(vki)
#koşulları yazdırmak 8 paun
if vki < 18:
  print("Zayıf")
elif 18 <= vki <= 25:
  print("normal")
elif vki <= 30:
  print("kilolu")
elif vki <= 35:
  print("obez")
else:
  print("Aşırı Obez")
```

# **Soru 5**
fonksiyon yazınız kişinin adını, adı kaç harf ise o kadar kere yazsın ve geriye adının kaç harf olduğunu döndürsün

```python
def harf_kadar_yazdir(): #fonskiyonu tanımlamak 6 puan
  ad = input("Adınızı Giriniz")
  harf_sayisi = len(ad)
  print(ad*harf_sayisi)
  #yukarıdaki satırlar 5 puan
  return harf_sayisi #geri dönüş 5 puan

harf_kadar_yazdir()
```

# **soru 6**
tahmin(x) fonksiyonunu yazarak gelen değeri tahmin aralığında değerlendirerek tahminin tutup tutmadığını yazdırınız

```python
from random import randint
#kütüphane hadil etmek 4 puan
def tahmin(x):  # fonksiyon tanımlamak 4 puan
  tutulan = randint(1,10) # sayı üretmek 4 puan
  if x == tutulan:
    print("Doğru Tahmin")
  else:
    print("Hatalı Tahmin")
  print(tutulan)
  #yukarıdaki kodlar 4 puan
tahmin(5) # çağırmak 4 puan
```
